```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get the highest salary for each age
highest_salary_by_age = (
    df.groupby('Age')['Annual_Salary_LPA']
    .max()
    .reset_index()
)

# Create the bar plot
plt.figure(figsize=(10, 5))
sns.barplot(
    data=highest_salary_by_age,
    x='Age',
    y='Annual_Salary_LPA'
)

plt.title('Highest Salary by Age')
plt.xlabel('Age')
plt.ylabel('Highest Salary (LPA)')
plt.show()
```

```python
highest = highest_salary_by_age.loc[
    highest_salary_by_age['Annual_Salary_LPA'].idxmax()
]

print(highest)
```

```python
avg_salary_gender = df.groupby('Gender')['Annual_Salary_LPA'].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=avg_salary_gender,
            x='Gender',
            y='Annual_Salary_LPA')

plt.title('Average Salary by Gender')
plt.show()
```

```python
dept_salary = df.groupby('Department')['Annual_Salary_LPA'].mean().reset_index()

plt.figure(figsize=(12, 5))
sns.barplot(data=dept_salary,
            x='Department',
            y='Annual_Salary_LPA')

plt.xticks(rotation=45)
plt.title('Average Salary by Department')
plt.show()
```

```python
top_employees = df.nlargest(10, 'Annual_Salary_LPA')

plt.figure(figsize=(12, 5))
sns.barplot(data=top_employees,
            x='Experience_Years',
            y='Annual_Salary_LPA')

plt.xticks(rotation=45)
plt.title('Top 10 Highest Paid Employees')
plt.show()
```

```python
top_employees = df.nlargest(10, 'Annual_Salary_LPA')

plt.figure(figsize=(12, 5))
sns.barplot(data=top_employees,
            x='Skill_Score',
            y='Annual_Salary_LPA')

plt.xticks(rotation=45)
plt.title('Top 10 Highest Paid Employees')
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(8, 6))
sns.heatmap(numeric_cols.corr(),
            annot=True,
            cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```
